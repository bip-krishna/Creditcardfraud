## 1. Load and inspect dataset
We check the row/column counts, column types and first few transactions before making any changes.

```python
csv_path = Path("credit_card_fraud_10k.csv")
if not csv_path.exists():
    raise FileNotFoundError("Put credit_card_fraud_10k.csv in the same folder as the notebook.")
df = pd.read_csv(csv_path)
print("Rows and columns:", df.shape)
display(df.head())
print("Data types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
display(df.isna().sum().to_frame("missing_count"))
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate transaction_id values:", df["transaction_id"].duplicated().sum())
print("Fraud label counts:")
display(df["is_fraud"].value_counts().rename(index={0:"Legitimate",1:"Fraudulent"}).to_frame("count"))
```

**Observation (verified on the supplied CSV):** 10,000 transactions, 10 columns, zero missing cells, zero duplicate rows and zero duplicate transaction IDs. There are **151 fraud-labeled transactions (1.51%)** and 9,849 legitimate transactions (98.49%). The classes are imbalanced, so compare **fraud rates within groups**, not fraud counts alone.

## 2. Basic cleaning and data validation
- Retain **all original CSV column names**.
- Check nulls, duplicates, types, and reasonable value ranges.
- Do not delete amount outliers automatically: large values may be relevant to this descriptive study.
- Check whether any actual correction is needed rather than pretending we cleaned errors that were absent.

```python
numeric_cols = [col for col in df.columns if col != "merchant_category"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="raise")

df = df.drop_duplicates().copy()
checks = {
    "unique_transaction_id": df["transaction_id"].is_unique,
    "valid_fraud_labels": df["is_fraud"].isin([0, 1]).all(),
    "valid_hours": df["transaction_hour"].between(0, 23).all(),
    "valid_binary_flags": df["foreign_transaction"].isin([0, 1]).all() and df["location_mismatch"].isin([0, 1]).all(),
    "nonnegative_amounts": df["amount"].ge(0).all(),
    "valid_trust_score": df["device_trust_score"].between(0, 100).all(),
    "nonnegative_velocity": df["velocity_last_24h"].ge(0).all(),
    "nonnegative_age": df["cardholder_age"].ge(0).all(),
}
display(pd.DataFrame(checks.items(), columns=["Check", "Passed?"]))
assert all(checks.values()), "Some values require inspection."
print("Final rows after checks:", len(df))
```

```python
q1 = df["amount"].quantile(0.25)
q3 = df["amount"].quantile(0.75)
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr
outlier_count = (df["amount"] > upper_fence).sum()
print(f"Amount upper IQR fence: {upper_fence:.2f}")
print("Transactions flagged above fence:", outlier_count)
print("Action: keep them and mention their effect on the amount plots.")
```

## 3. 3NF SQLite database design (inside Jupyter)
The dataset lacks a reliable `cardholder_id`; therefore, we **do not invent a cardholder table**. Our tables are:

1. `merchant_categories` - one unique `merchant_category` per row.
2. `transactions` - transaction identifier, amount, hour, category foreign key, age, fraud label.
3. `transaction_indicators` - one-to-one transaction indicators, linked by `transaction_id`.

Each non-key column depends on the key of its table, and no non-key field is stored redundantly between these final tables. This design is in **3NF**. Separating indicators is a logical organization choice, **not a requirement of 3NF** itself. The `transactions_raw` table below is only a **temporary staging table** and is dropped after loading.

**All SQLite commands below are executed inside Jupyter with `%sql` / `%%sql`.**

```python
%load_ext sql
```

### 3.1 Create related tables
For a clean rerun, delete only the three project tables before recreating them. Primary/foreign keys and `CHECK` rules protect the data.

```python
%%sql
PRAGMA foreign_keys = ON;
DROP TABLE IF EXISTS transaction_indicators;
DROP TABLE IF EXISTS transactions;
DROP TABLE IF EXISTS merchant_categories;
CREATE TABLE merchant_categories (
    merchant_category TEXT PRIMARY KEY
);
CREATE TABLE transactions (
    transaction_id INTEGER PRIMARY KEY,
    amount REAL NOT NULL CHECK (amount >= 0),
    transaction_hour INTEGER NOT NULL CHECK (transaction_hour BETWEEN 0 AND 23),
    merchant_category TEXT NOT NULL REFERENCES merchant_categories(merchant_category),
    cardholder_age INTEGER NOT NULL CHECK (cardholder_age >= 0),
    is_fraud INTEGER NOT NULL CHECK (is_fraud IN (0, 1))
);
CREATE TABLE transaction_indicators (
    transaction_id INTEGER PRIMARY KEY REFERENCES transactions(transaction_id),
    foreign_transaction INTEGER NOT NULL CHECK (foreign_transaction IN (0, 1)),
    location_mismatch INTEGER NOT NULL CHECK (location_mismatch IN (0, 1)),
    device_trust_score INTEGER NOT NULL CHECK (device_trust_score BETWEEN 0 AND 100),
    velocity_last_24h INTEGER NOT NULL CHECK (velocity_last_24h >= 0)
);
```

```python
%%sql
INSERT INTO merchant_categories (merchant_category)
SELECT DISTINCT merchant_category FROM transactions_raw;
INSERT INTO transactions (
    transaction_id, amount, transaction_hour, merchant_category,
    cardholder_age, is_fraud
)
SELECT transaction_id, amount, transaction_hour, merchant_category,
       cardholder_age, is_fraud
FROM transactions_raw;
INSERT INTO transaction_indicators (
    transaction_id, foreign_transaction, location_mismatch,
    device_trust_score, velocity_last_24h
)
SELECT transaction_id, foreign_transaction, location_mismatch,
       device_trust_score, velocity_last_24h
FROM transactions_raw;
DROP TABLE transactions_raw;
```

### 3.2 Demonstrate the relational structure and successful load

```python
%%sql
SELECT 'merchant_categories' AS table_name, COUNT(*) AS row_count FROM merchant_categories
UNION ALL
SELECT 'transactions', COUNT(*) FROM transactions
UNION ALL
SELECT 'transaction_indicators', COUNT(*) FROM transaction_indicators;
```

```python
%%sql
PRAGMA foreign_key_check;
```

**Interpretation:** Expected counts are 5 merchant categories, 10,000 transactions and 10,000 corresponding indicator records. `PRAGMA foreign_key_check` should return no violations.

```python
%%sql
SELECT t.transaction_id, t.merchant_category, t.amount,
       i.foreign_transaction, i.location_mismatch, i.device_trust_score,
       t.is_fraud
FROM transactions t
JOIN merchant_categories m ON t.merchant_category=m.merchant_category
JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
ORDER BY t.transaction_id
LIMIT 8;
```

## 4. EDA with SQL queries and plots
For comparisons, **fraud rate (%) = 100 × fraud transactions / total transactions in that group**. Always show the denominator because rates based on very small groups can be misleading.

### 4.1 Distribution of fraudulent and legitimate transactions

```python
%%sql
SELECT is_fraud, COUNT(*) AS total_transactions,
       ROUND(100.0 * COUNT(*) / (SELECT COUNT(*) FROM transactions), 2) AS percentage
FROM transactions GROUP BY is_fraud ORDER BY is_fraud;
```

```python
counts = df["is_fraud"].value_counts().sort_index()
ax = counts.rename(index={0:"Legitimate",1:"Fraudulent"}).plot(kind="bar", figsize=(6,3))
ax.set(title="Fraud vs legitimate transactions", ylabel="Transaction count", xlabel="Label")
plt.xticks(rotation=0)
for i,n in enumerate(counts.values): ax.text(i,n+80,str(n),ha="center")
plt.tight_layout(); plt.show()
```

**Observation:** 151 of 10,000 transactions are labeled fraudulent (1.51%), versus 9,849 legitimate (98.49%). This class imbalance is why we use within-group rates below.

### 4.2 Merchant categories (SQL `JOIN` + `GROUP BY`)

```python
%%sql merchant_results <<
SELECT t.merchant_category, COUNT(*) AS total_transactions,
       SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN merchant_categories m
ON t.merchant_category=m.merchant_category
GROUP BY t.merchant_category ORDER BY fraud_rate_pct DESC;
```

```python
merchant_df = merchant_results.DataFrame()
display(merchant_df)
ax = sns.barplot(data=merchant_df, x="merchant_category", y="fraud_rate_pct", color="steelblue")
ax.set(xlabel="Merchant category", ylabel="Observed fraud rate (%)", title="Fraud rate by merchant category")
plt.tight_layout(); plt.show()
```

**Observation:** In this dataset, Grocery has the highest observed rate among the five merchant categories (39/1,944 = 2.01%). Differences do not establish causation.

### 4.3 Transaction hour

```python
%%sql hour_results <<
SELECT transaction_hour, COUNT(*) AS total_transactions,
       SUM(is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions GROUP BY transaction_hour ORDER BY transaction_hour;
```

```python
hour_df = hour_results.DataFrame()
display(hour_df)
plt.figure(figsize=(8,3))
sns.lineplot(data=hour_df, x="transaction_hour", y="fraud_rate_pct", marker="o")
plt.xticks(range(0,24,2))
plt.title("Fraud rate by transaction hour")
plt.ylabel("Fraud rate (%)"); plt.xlabel("Hour (0-23)")
plt.tight_layout(); plt.show()
```

**Observation:** The dataset shows higher observed fraud rates from hours 0 to 3 (e.g., hour 0: 37/417 = 8.87%). Other hours have lower rates. The underlying data are synthetic.

### 4.4 Foreign transaction flag and location mismatch

```python
%%sql foreign_results <<
SELECT i.foreign_transaction, COUNT(*) AS total_transactions,
       SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
GROUP BY i.foreign_transaction ORDER BY i.foreign_transaction;
```

```python
%%sql location_results <<
SELECT i.location_mismatch, COUNT(*) AS total_transactions,
       SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
GROUP BY i.location_mismatch ORDER BY i.location_mismatch;
```

```python
foreign_df = foreign_results.DataFrame()
location_df = location_results.DataFrame()
print("Foreign transactions:"); display(foreign_df)
print("Location mismatch:"); display(location_df)
fig, axes = plt.subplots(1, 2, figsize=(9,3))
for a, z, column, title in [
    (axes[0], foreign_df, "foreign_transaction", "Foreign transaction"),
    (axes[1], location_df, "location_mismatch", "Location mismatch"),
]:
    sns.barplot(data=z, x=column, y="fraud_rate_pct", ax=a, color="steelblue")
    a.set(title=title, ylabel="Observed fraud rate (%)", xlabel="0 = No, 1 = Yes")
plt.tight_layout(); plt.show()
```

**Observation:** Foreign-transaction flag: 82/978 = 8.38% when 1, versus 69/9,022 = 0.76% when 0. Location mismatch: 72/857 = 8.40% when 1, versus 79/9,143 = 0.86% when 0. These are separate groupings and may overlap.

### 4.5 Transaction amount (outliers retained)

```python
%%sql amount_results <<
SELECT CASE WHEN amount < 50 THEN 'Below 50'
                 WHEN amount < 200 THEN '50 to below 200'
                 WHEN amount < 500 THEN '200 to below 500'
                 ELSE '500 or more' END AS amount_group,
       COUNT(*) AS total_transactions, SUM(is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions GROUP BY amount_group
ORDER BY CASE amount_group WHEN 'Below 50' THEN 1 WHEN '50 to below 200' THEN 2 WHEN '200 to below 500' THEN 3 ELSE 4 END;
```

```python
amount_df = amount_results.DataFrame()
display(amount_df)
fig, axes = plt.subplots(1, 2, figsize=(10,3))
sns.histplot(data=df, x="amount", hue="is_fraud", bins=35, ax=axes[0])
axes[0].set_title("Transaction amount distribution")
sns.boxplot(data=df, x="is_fraud", y="amount", ax=axes[1])
axes[1].set_xticklabels(["Legitimate", "Fraudulent"])
axes[1].set_title("Amount by transaction label")
plt.tight_layout(); plt.show()
```

**Observation:** Fraud transactions have higher mean amounts (216.18 versus 175.33), but their median (118.94) is similar to legitimate transactions (122.11). An unusually large amount is not itself proof of fraud.

### 4.6 Device trust score and transaction frequency

```python
%%sql device_results <<
SELECT CASE WHEN i.device_trust_score < 40 THEN 'Low (<40)'
                 WHEN i.device_trust_score < 70 THEN 'Medium (40-69)'
                 ELSE 'High (70-100)' END AS trust_group,
       COUNT(*) AS total_transactions, SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
GROUP BY trust_group ORDER BY CASE trust_group WHEN 'Low (<40)' THEN 1 WHEN 'Medium (40-69)' THEN 2 ELSE 3 END;
```

```python
%%sql velocity_results <<
SELECT CASE WHEN i.velocity_last_24h >= 5 THEN '5 or more' ELSE 'Below 5' END AS velocity_group,
       COUNT(*) AS total_transactions, SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
GROUP BY velocity_group ORDER BY velocity_group;
```

```python
device_df = device_results.DataFrame()
velocity_df = velocity_results.DataFrame()
print("Trust score bands:"); display(device_df)
print("Previous 24-hour transaction frequency:"); display(velocity_df)
fig, axes = plt.subplots(1, 2, figsize=(10,3))
sns.barplot(data=device_df, x="trust_group", y="fraud_rate_pct", ax=axes[0], color="steelblue")
axes[0].tick_params(axis="x", rotation=15)
axes[0].set(title="Device trust bands", ylabel="Fraud rate (%)", xlabel="Trust score band")
sns.barplot(data=velocity_df, x="velocity_group", y="fraud_rate_pct", ax=axes[1], color="steelblue")
axes[1].set(title="Recent transaction frequency", ylabel="Fraud rate (%)", xlabel="Number of recent transactions")
plt.tight_layout(); plt.show()
```

**Observation:** Fraud-labeled records have a lower mean device trust score (37.87 versus 62.17) and a higher mean `velocity_last_24h` (3.21 versus 1.99). This describes associations in the synthetic sample.

### 4.7 Age-group comparison

```python
%%sql age_results <<
SELECT CASE WHEN cardholder_age < 30 THEN '18 to 29'
                 WHEN cardholder_age < 45 THEN '30 to 44'
                 WHEN cardholder_age < 60 THEN '45 to 59'
                 ELSE '60 and above' END AS age_group,
       COUNT(*) AS total_transactions, SUM(is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions GROUP BY age_group
ORDER BY CASE age_group WHEN '18 to 29' THEN 1 WHEN '30 to 44' THEN 2 WHEN '45 to 59' THEN 3 ELSE 4 END;
```

```python
age_df = age_results.DataFrame()
display(age_df)
ax = sns.barplot(data=age_df, x="age_group", y="fraud_rate_pct", color="steelblue")
ax.set(xlabel="Age band", ylabel="Fraud rate (%)", title="Fraud rates by age group")
plt.tight_layout(); plt.show()
```

**Observation:** The mean cardholder age is approximately the same across fraud and legitimate transactions (~43.4 years). An age association should not be exaggerated.

### 4.8 Optional: two suspicious flags together

```python
%%sql combined_results <<
SELECT i.foreign_transaction, i.location_mismatch,
       COUNT(*) AS total_transactions, SUM(t.is_fraud) AS fraud_transactions,
       ROUND(100.0 * SUM(t.is_fraud) / COUNT(*), 2) AS fraud_rate_pct
FROM transactions t JOIN transaction_indicators i ON t.transaction_id=i.transaction_id
GROUP BY i.foreign_transaction,i.location_mismatch ORDER BY i.foreign_transaction,i.location_mismatch;
```

```python
combined_df = combined_results.DataFrame()
display(combined_df)
```

**Observation:** The combined table lets us compare four non-overlapping groups. Always consider the group sizes alongside rates.

### 4.9 Correlation heatmap

## 5. Summary, limitations and reflections
**Findings in the provided synthetic sample:**

- Dataset fraud prevalence is **1.51% (151/10,000)**. Rates must be shown alongside group sizes.
- Higher observed rates occur for foreign transactions and location mismatches (~8.4% each for the flagged groups).
- Fraud-labeled records are more frequent in the midnight to 3 a.m. hours in this dataset.
- Lower device trust scores and higher recent transaction frequency are associated with the fraud label.
- Merchant category differences exist, but are smaller than the differences seen in some other groupings.

**Limitations:** These are simulated records, so patterns may reflect how synthetic examples were generated. We do not train a classifier, infer causal relationships, or claim a practical fraud detection system. Age and other features should not be used to make personal judgments about individuals.

## 6. Viva preparation (quick answers)
**Q: Why SQL instead of Pandas alone?** The course requires a normalized relational SQLite database and querying it from Python/Jupyter.  
**Q: What is 3NF?** A table is in 3NF when it is in 2NF and has no transitive dependency of a non-key attribute on a candidate key; here each table's descriptive fields depend on its own key without another non-key field determining them.  
**Q: Why do you join on `transaction_id`?** It is the unique transaction key; `transaction_indicators.transaction_id` is both primary key and foreign key.  
**Q: Why compare rates instead of counts?** Group sizes differ, so counts alone are misleading.  
**Q: Why not remove all outliers?** Unusually large transaction amounts might be informative; we flag and discuss them.  
**Q: Did you predict fraud?** No. The goal is descriptive EDA, as stated in the abstract.  
**Q: Are conclusions about real financial fraud justified?** No. The source is synthetic.
